```python
from dotenv import load_dotenv
load_dotenv(".env")

from sklearn.datasets import load_iris
import numpy as np
import numpyai_dashboard as npi

from rich.console import Console
```

```python
c = Console() 
```

```python
iris = load_iris()
X, y = iris.data, iris.target
print(isinstance(X, np.ndarray) and isinstance(y, np.ndarray))
```

```python
sess = npi.NumpyAISession([X, y], verbose=False)
diag = npi.Diagnosis(sess)
```

```python
steps = diag.steps(task="Give me the exact and pithy steps to decide which machine learning model to use. Don't go into the Machine Learning side, rather use simple data-analysis. You should not give more than seven steps.")
```

```python
for step in steps:
    try:
        vals = sess.chat(step)
        print(vals)
    except:
        c.print(f"The following step failed: {step}")
```
